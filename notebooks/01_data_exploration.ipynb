```python
import json
from pathlib import Path

import pandas as pd
```

```python
DATA_PATH = Path("../data/raw/adzuna_jobs.json")

with open(DATA_PATH, "r", encoding="utf-8") as f:
    jobs = json.load(f)
print(f"Number of jobs: {len(jobs)}")
print(f"Type of object: {type(jobs)}")
```

```python
#Inspect the first job posting
sample_job = jobs[0]

print(f"Type: {type(sample_job)}")
print(f"Number of keys: {len(sample_job)}")
print(f"\nKeys available:")
for key in sample_job.keys():
    print(f"- {key}")
    
```

```python
print(json.dumps(sample_job, indent=2, ensure_ascii=False))
```

```python
#Flattent the nested JSON data into a pandas DataFrame
df = pd.json_normalize(jobs)

print(f"Shape: {df.shape}")
print(f"Columns: ({len(df.columns)})")
for col in df.columns:
    print(f"- {col}")
```

```python
df.head(3)
```

```python
#show column types and non-null counts for each column
df.info()
```

```python
df.isnull().sum().sort_values(ascending=False)
```

```python
df.describe()
```

```python
# List of colmns we don't need for analysis
columns_to_drop = [
    "__CLASS__",
    "company.__CLASS__",
    "category.__CLASS__",
    "location.__CLASS__",
    "adref",
    "salary_is_predicted",
]
# Drop the columns
df = df.drop(columns=columns_to_drop)
print(f"New shape: {df.shape}") 
print(f"Remaining columns:{list(df.columns)}")
```

```python
# Rename nested to cleaner names
df = df.rename(
    columns={
        "company.display_name": "company",
        "category.label": "category",
        "category.tag": "category_tag",
        "location.display_name": "location",   
        "location.area": "location_area",

    }
)
print(f"Colums after renaming:")
for col in df.columns:
    print(f"- {col}")
```

```python
#Inspect the firs 5 location_area values to confirm the structure
for i in range(5):
    print(f"Job {i} :{df['location_area'].iloc[i]}")
```

```python
#Extract the city (index 2 of the location_area list) 
df["city"] = df["location_area"].apply(lambda area : area[2] if len(area) > 2 else None)
#Check the result
print(df[["location_area", "city"]].head(10))
```

```python
#Count the most common cities in the dataset
df["city"].value_counts().head(15)
```

```python
# Inspect the created columns to understand its format
print(f"Type of the column : {df['created'].dtype }")
print(f"\nFirst 5 values:")
print(df["created"].head())
```

```python
# Convert the data string to a real datetime type
df["created"] = pd.to_datetime(df["created"])
#Confirm the new type and inspect the values
print(f"Type of the column : {df['created'].dtype }")
print(f"\nDate range: from {df['created'].min()} to {df['created'].max()}")
print(f"Total span : {(df['created'].max() - df['created'].min())}")
```

```python
#Add week and day of week columns for later analysis
df["week"] = df["created"].dt.to_period("W").astype(str)
df["day_of_week"] = df["created"].dt.day_name()
print(df[["created", "week", "day_of_week"]].head())
```

```python
#Outlier cleaning for salary
# Replace unrealistic salaries with NaN (anything below 25,000 CAD is suspicious)
SALARY_THRESHOLD = 25000

df.loc[df["salary_min"] < SALARY_THRESHOLD, "salary_min"] = None
df.loc[df["salary_max"] < SALARY_THRESHOLD, "salary_max"] = None

# Verify
print("Salary stats after cleaning:")
print(df[["salary_min", "salary_max"]].describe())
```

```python
# Distribution of jobs by year
print("Jobs by year:")
print(df["created"].dt.year.value_counts().sort_index())

print("\nJobs by year-month (last 12 months only):")
print(df["created"].dt.to_period("M").value_counts().sort_index().tail(12))
```

```python
# Flag jobs from the active market window (2026 only)
df["is_recent"] = df["created"].dt.year == 2026

# Quick check
print(f"Recent jobs (2026): {df['is_recent'].sum()}")
print(f"Older jobs: {(~df['is_recent']).sum()}")
```

```python
OUTPUT_PATH = Path("../data/processed/adzuna_jobs_cleaned.csv")
df.to_csv(OUTPUT_PATH, index=False, encoding="utf-8")

print(f"Saved {len(df)} rows to {OUTPUT_PATH}")
print(f"File size :{OUTPUT_PATH.stat().st_size / (1024 * 1024):.1f} KB")
```
